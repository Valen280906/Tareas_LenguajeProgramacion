## 📘 Descripción
Ruby es un lenguaje de programación interpretado, orientado a objetos y con una sintaxis muy legible. Fue creado por Yukihiro Matsumoto ('Matz') en 1995 y es conocido por su elegancia y simplicidad.

```

```

## 💡 Ejemplos de sintaxis básica
En esta sección se muestran ejemplos sobre variables, tipos de datos y operaciones básicas.

````
# Variables en Ruby
nombre = "Valentina"
edad = 25
altura = 1.68
es_estudiante = true

puts "Hola, mi nombre es #{nombre}, tengo #{edad} años y mido #{altura} m."
puts "¿Soy estudiante? #{es_estudiante}"
````

````
# Tipos de datos y operaciones aritméticas
a = 10
b = 3

puts "Suma: #{a + b}"
puts "Resta: #{a - b}"
puts "Multiplicación: #{a * b}"
puts "División: #{a / b.to_f}"
puts "Módulo: #{a % b}"
puts "Potencia: #{a ** b}"
````

````
# Booleanos y operadores lógicos
x = true
y = false

puts "x AND y => #{x && y}"
puts "x OR y  => #{x || y}"
puts "NOT x   => #{!x}"
````

## 🔁 Estructuras de control
Condicionales, bucles y control de flujo.

````
# Condicional if/elsif/else
edad = 20

if edad < 18
  puts "Eres menor de edad"
elsif edad >= 18 && edad < 65
  puts "Eres adulto"
else
  puts "Eres adulto mayor"
end
````

````
# Bucles
puts "\nBucle while:"
i = 1
while i <= 5
  puts i
  i += 1
end

puts "\nBucle for:"
for num in 1..5
  puts num
end
````

## 🧮 Funciones
Ruby define funciones con la palabra clave `def`.

````
def sumar(a, b)
  return a + b
end

resultado = sumar(4, 7)
puts "El resultado de la suma es: #{resultado}"
````

## 🧱 Programación Orientada a Objetos (POO)
Ruby es 100% orientado a objetos. Todo es un objeto, incluso los números y cadenas.

````
class Persona
  attr_accessor :nombre, :edad

  def initialize(nombre, edad)
    @nombre = nombre
    @edad = edad
  end

  def saludar
    puts "Hola, soy #{@nombre} y tengo #{@edad} años."
  end
end

persona1 = Persona.new("Valentina", 25)
persona1.saludar
````

## 🧰 Vectores (Arreglos) y sus métodos
Los arreglos en Ruby son muy flexibles y pueden contener distintos tipos de datos.

````
numeros = [10, 20, 30, 40, 50]

puts "Arreglo original: #{numeros}"
puts "Primer elemento: #{numeros.first}"
puts "Último elemento: #{numeros.last}"
puts "Tamaño: #{numeros.length}"

# Métodos útiles
numeros.push(60)
puts "Después de push: #{numeros}"

numeros.pop
puts "Después de pop: #{numeros}"

numeros.each { |n| puts "Elemento: #{n}" }
````
