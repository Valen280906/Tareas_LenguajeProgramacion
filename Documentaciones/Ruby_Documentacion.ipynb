
# Documentación de Ruby

## 1. Introducción

Ruby es un **lenguaje de programación interpretado, de alto nivel y orientado a objetos**, creado en **1993** por **Yukihiro "Matz" Matsumoto** en Japón.  
Su creador buscaba un lenguaje **simple y elegante**, que combinara el poder de Perl con la claridad de Smalltalk.  
El lema de Ruby es: *“Optimized for developer happiness”* —es decir, está diseñado para que **programar sea una experiencia placentera y natural**.

### Tipo de lenguaje
- Interpretado y dinámico  
- Tipado dinámico y fuerte  
- Altamente orientado a objetos (todo en Ruby es un objeto, incluso los números)  

### Usos comunes
- Desarrollo web (especialmente con el framework **Ruby on Rails**)  
- Automatización de tareas  
- Scripting en servidores  
- Prototipado rápido y aplicaciones empresariales  

**Dato curioso:** Ruby fue diseñado con una fuerte influencia filosófica: su prioridad es el **humano, no la máquina**. Matz dijo una vez:  
> “Quería hacer un lenguaje que hiciera feliz a los programadores.”



## 2. Instalación y Entorno de Ejecución

Para comenzar con Ruby, necesitamos un entorno donde ejecutar y probar el código.  
A continuación, se describen algunas opciones comunes para instalarlo y configurarlo:

### Opción 1: Instalación directa
- En Windows: descarga **RubyInstaller** desde [rubyinstaller.org](https://rubyinstaller.org/).  
- En macOS o Linux: la mayoría ya incluye Ruby preinstalado, pero puede actualizarse con:  
  ```bash
  sudo apt install ruby-full
  ```




## 3. Sintaxis básica de un script Ruby

Un archivo Ruby tiene la extensión `.rb`. Su sintaxis es **limpia y expresiva**, sin necesidad de llaves `{}` ni punto y coma `;` al final de cada línea.

```ruby
# Hola Mundo en Ruby
puts "¡Hola Mundo!"

# Comentarios en Ruby
# Esto es un comentario de una sola línea

=begin
Esto es un comentario
de varias líneas
=end
```

Ruby usa `puts` o `print` para mostrar texto en consola. La diferencia es que `puts` agrega un salto de línea automáticamente.



## 4. Estructuras Repetitivas

Ruby ofrece diversas formas de repetir acciones. Los bucles son una parte fundamental del lenguaje.

### 4.1 While
Ejecuta un bloque de código **mientras** la condición sea verdadera.

```ruby
contador = 1
while contador <= 5
  puts "Iteración número #{contador}"
  contador += 1
end
```

### 4.2 Until
Hace lo contrario: repite el bloque **hasta que** la condición sea verdadera.

```ruby
numero = 0
until numero == 3
  puts "Número actual: #{numero}"
  numero += 1
end
```

### 4.3 For
Itera sobre un rango o una colección.

```ruby
for i in 1..5
  puts "El valor de i es #{i}"
end
```

### 4.4 Each
Ruby prefiere la forma más “humana” de iterar: usando `each`, ideal para recorrer arrays o hashes.

```ruby
[1, 2, 3, 4].each do |num|
  puts "Número: #{num}"
end
```

**Dato:** En Ruby, los bucles son expresiones, lo que significa que pueden devolver valores.



## 5.5. Funciones (Métodos)

En Ruby, las funciones se definen con la palabra clave `def` y terminan con `end`.  
Todas las funciones devuelven automáticamente el valor de la última expresión evaluada (no es necesario `return`, aunque puede usarse).

```ruby
def saludar(nombre)
  "Hola, #{nombre}! Bienvenido a Ruby."
end

puts saludar("Isabella")
```

### Parámetros por defecto
```ruby
def despedir(nombre = "visitante")
  "Adiós, #{nombre}."
end

puts despedir
puts despedir("Valentina")
```

**Nota:** Los métodos en Ruby son **ciudadanos de primera clase**, lo que significa que pueden pasarse como argumentos, asignarse a variables y devolverse como resultados.



## 6.6. Manejo de Errores

Ruby maneja los errores mediante **excepciones**, usando las palabras clave `begin`, `rescue`, `else`, y `ensure`.

```ruby
begin
  resultado = 10 / 0
rescue ZeroDivisionError => e
  puts "Error: #{e.message}"
else
  puts "Operación exitosa."
ensure
  puts "Bloque 'ensure' ejecutado siempre."
end
```

**Dato:** `ensure` se ejecuta **siempre**, ocurra o no una excepción, ideal para cerrar archivos o liberar recursos.



## 7.. Arreglos y Hashes

### Arreglos (Arrays)
Los arrays en Ruby son **colecciones ordenadas** de elementos.

```ruby
frutas = ["manzana", "banana", "cereza"]
puts frutas[0]           # Accede al primer elemento
frutas << "mango"         # Agrega un elemento
puts frutas.join(", ")    # Une los elementos con comas
```

### Hashes (diccionarios)
Un Hash almacena **pares clave-valor**, parecido a un diccionario en otros lenguajes.

```ruby
persona = { "nombre" => "Isabella", "edad" => 20, "profesion" => "Estudiante" }

puts persona["nombre"]
persona["edad"] = 21
puts persona
```

**Dato:** Desde Ruby 1.9, puedes usar símbolos como claves:  
```ruby
usuario = { nombre: "Valen", activo: true }
puts usuario[:nombre]
```
