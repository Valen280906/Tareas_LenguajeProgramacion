# Documentación de PHP

## 1. Introducción

PHP (acrónimo recursivo de *PHP: Hypertext Preprocessor*) es un **lenguaje de programación del lado del servidor** creado en **1994** por **Rasmus Lerdorf**, un programador danés-canadiense que, en principio, solo deseaba registrar las visitas a su página personal.  
Aquel pequeño conjunto de scripts escritos en C, llamado originalmente *Personal Home Page Tools*, terminó convirtiéndose en uno de los pilares de la programación web moderna.

Mientras **JavaScript** actúa en el **navegador (lado del cliente)**, **PHP se ejecuta en el servidor**.  
Esto significa que el servidor interpreta el código PHP, genera el contenido dinámico (por ejemplo, al conectarse a una base de datos o procesar formularios) y luego envía al navegador el resultado final, que suele ser **HTML puro**.  
Por ello, el usuario nunca ve el código PHP, solo el contenido que este produce.

---

## 2. Tipo de lenguaje

- **Interpretado:** No requiere compilación previa. El servidor lee y ejecuta el código directamente.  
- **Tipado débil:** Las variables no necesitan declarar tipo; el lenguaje realiza conversiones automáticas según el contexto.  
- **Multiparadigma:** Desde PHP 5 incorporó la **orientación a objetos**, con soporte para clases, herencia, interfaces y excepciones.  
- **Multiplataforma:** Funciona en Windows, macOS, Linux y otros sistemas operativos.

---

## 3. Usos comunes

A pesar de su antigüedad, PHP continúa siendo una de las herramientas más utilizadas para construir la web.  
Su combinación de simplicidad, potencia y comunidad activa le ha permitido evolucionar junto a Internet.  
Algunos de sus usos más frecuentes son:

- Desarrollo de **sistemas web dinámicos** y **CMS** (como WordPress, Drupal o Joomla).  
- **Conexión con bases de datos** (MySQL, PostgreSQL, SQLite, entre otras).  
- Creación de **APIs REST** y servicios del lado del servidor.  
- Desarrollo de **tiendas en línea**, foros, blogs y plataformas educativas.  
- Integración con HTML, CSS y JavaScript para aplicaciones de tipo **full stack**.

---

## 4. Curiosidades históricas

En sus primeros años, PHP no era más que un conjunto de scripts CGI escritos en C.  
Con el tiempo, se transformó en un lenguaje con su propio motor de interpretación: **Zend Engine**.  
En el año **2000**, la llegada de **PHP 4** representó un salto importante en rendimiento y estabilidad.  
Con **PHP 5**, el lenguaje adoptó una estructura moderna orientada a objetos y consolidó su reputación.  
Hoy en día, PHP sigue siendo el corazón de sistemas como **WordPress**, **Wikipedia**, **Facebook (en sus inicios)** y una enorme cantidad de sitios web alrededor del mundo.

---

## 5. En resumen

PHP nació de la sencillez, creció junto con la web y maduró hasta convertirse en un lenguaje robusto y versátil.  
Aunque algunos lo consideran “clásico”, su presencia sigue siendo dominante: gran parte del contenido que circula por Internet pasa, en algún punto, por un servidor que ejecuta PHP.

---

> **Dato curioso:** El elefante de PHP, su mascota oficial, representa la fuerza, la memoria y la evolución constante del lenguaje a lo largo de las décadas.

---

## 6. Instalación y entorno de ejecución

Para ejecutar código PHP no basta con tener un editor de texto.  
Recordemos que es un **lenguaje del lado del servidor**, lo cual implica que necesita un entorno que simule o contenga un servidor web.  
Existen diversas maneras de hacerlo, pero una de las más sencillas para entornos de aprendizaje y desarrollo local es mediante **XAMPP**.

### 6.1 ¿Qué es XAMPP?

**XAMPP** es un paquete que integra varios componentes necesarios para ejecutar PHP en un entorno local, sin necesidad de configurar cada servicio de forma individual.  
El nombre proviene de sus partes:

- **X:** indica que funciona en distintos sistemas operativos (Windows, macOS, Linux).  
- **A:** Apache (servidor web).  
- **M:** MySQL o MariaDB (gestor de bases de datos).  
- **P:** PHP (el intérprete del lenguaje).  
- **P:** Perl (otro lenguaje de scripting, opcional).

En pocas palabras, XAMPP ofrece un servidor local “todo en uno”, ideal para pruebas, desarrollo y aprendizaje.

### 6.2 Instalación de PHP mediante XAMPP

1. **Descargar XAMPP:**  
   Dirígete al sitio oficial de Apache Friends ([https://www.apachefriends.org](https://www.apachefriends.org)) y descarga la versión compatible con tu sistema operativo.

2. **Instalar:**  
   Ejecuta el instalador y selecciona los componentes que deseas incluir.  
   Generalmente se recomienda mantener seleccionados **Apache**, **MySQL** y **PHP**.

3. **Iniciar servicios:**  
   Abre el **Panel de Control de XAMPP** y activa los módulos **Apache** (servidor web) y **MySQL** (base de datos).  
   Si las luces cambian a color verde, significa que están en ejecución correctamente.

4. **Crear tu primer proyecto:**  
   - Dentro del directorio de instalación (por ejemplo, `C:\xampp\htdocs` en Windows), crea una carpeta para tu proyecto.  
   - Dentro de ella, crea un archivo llamado `index.php`.  
   - Escribe en su interior:
     ```php
     <?php
     echo "¡Hola, mundo desde PHP!";
     ?>
     ```
   - Abre un navegador y escribe en la barra de direcciones:  
     `http://localhost/tu_carpeta/index.php`

5. **Verificar la instalación:**  
   Si aparece el mensaje en pantalla, significa que tu entorno PHP está funcionando correctamente.

---

### 6.3 ¿Por qué usar XAMPP?

- **Simplicidad:** Todo se instala en pocos pasos y sin configuraciones complicadas.  
- **Comodidad:** No requiere conexión a Internet ni servidores externos para ejecutar tus proyectos.  
- **Versatilidad:** Permite trabajar con bases de datos, scripts PHP y HTML en un mismo entorno.  
- **Ideal para estudiantes y desarrolladores principiantes:** Facilita la experimentación y comprensión de cómo funciona un servidor local antes de pasar a entornos en producción.

---

En resumen, **XAMPP** es el puente perfecto entre la teoría y la práctica.  
Permite comprender cómo se comporta PHP en un entorno real, pero dentro de la seguridad de un entorno local controlado.

---

> **Nota:** Aunque XAMPP es excelente para aprender y desarrollar, **no se recomienda para servidores en producción** por motivos de seguridad.  
> En entornos profesionales se utilizan configuraciones personalizadas o herramientas como **Docker**, **Nginx** o servidores dedicados de **Apache**.

## 7. Sintaxis básica de un script PHP

PHP es un lenguaje que puede integrarse directamente dentro del HTML.  
Esto significa que el servidor web interpreta el código PHP antes de enviar la página al navegador, generando contenido dinámico de acuerdo con las instrucciones del programador.

En esta sección exploraremos su estructura fundamental, la forma de declarar variables, los tipos de datos más comunes y las principales sentencias de salida.

---

### 7.1 Estructura mínima de un archivo PHP

Todo script en PHP comienza con la etiqueta `<?php` y termina con `?>`.  
Aunque esta etiqueta de cierre es opcional, se recomienda omitirla cuando el archivo contiene únicamente código PHP.

---

### 7.2 PHP incrustado en HTML

Una de las características más distintivas de PHP es que puede mezclarse con HTML.  
Esto permite generar secciones dinámicas, como mostrar la fecha actual, información de un usuario o resultados provenientes de una base de datos.

---

### 7.3 Comentarios

Los comentarios no son ejecutados por el intérprete y se utilizan para dejar notas o explicaciones dentro del código.  
Existen tres formas de escribirlos: con `//`, `#` o con `/* ... */` para varias líneas.

---

### 7.4 Variables y tipado

Todas las variables en PHP comienzan con el símbolo `$`.  
El lenguaje no exige declarar su tipo; PHP determina automáticamente si es una cadena, número, booleano, etc.  
Es importante recordar que distingue entre mayúsculas y minúsculas.

---

### 7.5 Tipos de datos básicos

PHP reconoce varios tipos primitivos de datos:

- **String:** texto o secuencia de caracteres.  
- **Integer:** número entero.  
- **Float:** número con decimales.  
- **Boolean:** valores lógicos `true` o `false`.  
- **Array:** conjunto ordenado de valores.  
- **Object:** instancia de una clase.  
- **NULL:** valor vacío o sin asignar.

---

### 7.6 Constantes

Las constantes almacenan valores fijos que no pueden modificarse.  
Se definen con la función `define()` o mediante la palabra clave `const`.

---

### 7.7 Sentencias de salida

Para mostrar información en pantalla, PHP ofrece varias funciones como `echo`, `print`, `print_r()` y `var_dump()`.  
Cada una tiene usos específicos según el nivel de detalle o el tipo de dato que se desea mostrar.

---

### 7.8 Buenas prácticas iniciales

1. Usar nombres de variables descriptivos y claros.  
2. Mantener la indentación del código.  
3. Evitar etiquetas cortas `<?= ?>` cuando sea posible.  
4. Comentar secciones importantes del código.  
5. Separar la lógica (PHP) del diseño (HTML) siempre que sea posible.

---

En resumen, la sintaxis básica de PHP es sencilla y flexible, ideal para aprender la lógica del lado del servidor.  
Con solo unas pocas líneas, es posible generar páginas dinámicas, procesar información y construir la base de aplicaciones web completas.


````php
<?php
$nombre = 'Stella';
$edad = 22;
$promedio = 18.7;
$es_activa = true;
echo "Nombre: $nombre\n";
echo "Edad: $edad\n";
echo "Promedio: $promedio\n";
echo "Activa: $es_activa\n";
?>
````

````php
<?php
// --------------------------
// Ejemplo 1: Script básico
// --------------------------

echo "Hola, mundo desde PHP!";

// --------------------------
// Ejemplo 2: PHP incrustado en HTML
// --------------------------
?>

<!DOCTYPE html>
<html>
<head>
    <title>Ejemplo PHP + HTML</title>
</head>
<body>
    <h1>Bienvenido</h1>
    <p>
        Hoy es:
        <?php echo date("d/m/Y"); ?>
    </p>
</body>
</html>

<?php
// --------------------------
// Ejemplo 3: Comentarios
// --------------------------

// Este es un comentario de una sola línea
# También válido para una línea
/*
   Comentario de varias líneas
   Ideal para explicar bloques largos de código
*/

// --------------------------
// Ejemplo 4: Variables y concatenación
// --------------------------

$nombre = "Isabella";
$edad = 19;
$estudiante = true;

echo "Nombre: " . $nombre . "<br>";
echo "Edad: " . $edad . " años<br>";
echo "¿Es estudiante?: " . ($estudiante ? "Sí" : "No") . "<br>";

// --------------------------
// Ejemplo 5: Tipos de datos
// --------------------------

$altura = 1.68;
$colores = ["rojo", "azul", "verde"];
$valor_nulo = null;

echo "<br>Altura: $altura m<br>";
print_r($colores);
echo "<br>";
var_dump($valor_nulo);

// --------------------------
// Ejemplo 6: Constantes
// --------------------------

define("PI", 3.1416);
const UNIVERSIDAD = "UNESR";

echo "<br>Valor de PI: " . PI . "<br>";
echo "Universidad: " . UNIVERSIDAD . "<br>";
?>
````

## 9. Operaciones aritméticas en PHP

> **Concepto general:**  
> PHP permite realizar operaciones matemáticas básicas y avanzadas utilizando operadores aritméticos.  
> Estos operadores trabajan con variables numéricas (enteros o flotantes) y son fundamentales para el cálculo de datos, el manejo de estadísticas o cualquier proceso lógico dentro de una aplicación web.

---

### 9.1 Operadores aritméticos

| Operador | Descripción | Ejemplo | Resultado |
|-----------|--------------|----------|------------|
| `+` | Suma | `$a + $b` | Suma de `a` y `b` |
| `-` | Resta | `$a - $b` | Resta de `a` y `b` |
| `*` | Multiplicación | `$a * $b` | Producto de `a` y `b` |
| `/` | División | `$a / $b` | Cociente entre `a` y `b` |
| `%` | Módulo o residuo | `$a % $b` | Resto de la división |
| `**` | Exponenciación | `$a ** $b` | `a` elevado a `b` |

---

### 9.2 Precedencia y agrupación

PHP evalúa las operaciones siguiendo las reglas de precedencia estándar (igual que en matemáticas).  
La multiplicación y la división se ejecutan antes que la suma o la resta, a menos que se usen paréntesis para modificar el orden:

```php
$resultado = (3 + 2) * 4;  // Resultado: 20


````php

<?php
// ------------------------------------
// Ejemplo 1: Operaciones básicas
// ------------------------------------
$a = 10;
$b = 3;

echo "Suma: " . ($a + $b) . "<br>";
echo "Resta: " . ($a - $b) . "<br>";
echo "Multiplicación: " . ($a * $b) . "<br>";
echo "División: " . ($a / $b) . "<br>";
echo "Módulo: " . ($a % $b) . "<br>";
echo "Potencia: " . ($a ** $b) . "<br>";

// ------------------------------------
// Ejemplo 2: Precedencia de operadores
// ------------------------------------
$resultado1 = 3 + 4 * 2;
$resultado2 = (3 + 4) * 2;

echo "<br>Sin paréntesis: $resultado1<br>";
echo "Con paréntesis: $resultado2<br>";

// ------------------------------------
// Ejemplo 3: Incremento y decremento
// ------------------------------------
$x = 5;

echo "<br>Valor inicial: $x<br>";
echo "Pre-incremento (++x): " . (++$x) . "<br>"; // aumenta antes
echo "Post-incremento (x++): " . ($x++) . "<br>"; // muestra y luego aumenta
echo "Valor final: $x<br>";

echo "Pre-decremento (--x): " . (--$x) . "<br>";
echo "Post-decremento (x--): " . ($x--) . "<br>";
echo "Valor final tras post-decremento: $x<br>";

// ------------------------------------
// Ejemplo 4: Conversión automática
// ------------------------------------
$texto = "10 manzanas";
$calculo = $texto + 5; // PHP tomará el número inicial del texto
echo "<br>Resultado de conversión automática: $calculo<br>";
?>
````

## 10. Estructuras de Control (Condicionales)

En todo lenguaje de programación, las **estructuras de control** son los mecanismos que permiten decidir qué camino tomará el programa según ciertas condiciones.  
En PHP, estas estructuras resultan esenciales para construir sitios web que **respondan dinámicamente** a la información del usuario o del servidor.

Podemos imaginarlas como una bifurcación en el camino: el código elige una dirección u otra dependiendo de si una afirmación es **verdadera (`true`)** o **falsa (`false`)**.

PHP, al ser un lenguaje de tipado débil, permite evaluar condiciones sin necesidad de declarar explícitamente los tipos de datos, lo que lo hace flexible, aunque requiere precaución para evitar resultados inesperados.

---

### 10.1. La estructura `if`

El condicional `if` evalúa una condición.  
Si dicha condición se cumple (es decir, si su valor lógico es `true`), el bloque de código asociado se ejecutará.  
De lo contrario, se ignora.



````php
<?php
$edad = 20;

if ($edad >= 18) {
    echo "Puedes ingresar, eres mayor de edad.";
}
?>
````

### 10.2. Estructura `if...else`

A veces, queremos ejecutar un bloque cuando la condición es verdadera y **otro distinto** cuando no lo es.  
Para ello, PHP ofrece la cláusula `else`, que se ejecuta si la condición no se cumple.

````php
<?php
$llueve = true;

if ($llueve) {
    echo "Lleva paraguas, está lloviendo.";
} else {
    echo "No necesitas paraguas, el cielo está despejado.";
}
?>
````

### 10.3. Estructura `if...elseif...else`

Cuando hay **más de dos caminos posibles**, podemos encadenar varias condiciones con `elseif`.  
El programa revisará una por una hasta encontrar la primera que sea verdadera.  
Si ninguna se cumple, ejecutará el bloque `else`.


````php
<?php
$hora = 14;

if ($hora < 12) {
    echo "Buenos días.";
} elseif ($hora < 18) {
    echo "Buenas tardes.";
} else {
    echo "Buenas noches.";
}
?>
````

### 10.4. Estructura `switch`

La estructura `switch` es útil cuando se desea comparar una misma variable contra múltiples valores posibles.  
Es más legible y ordenada que una serie larga de `if...elseif`.

> 💡 **Dato curioso:** En los inicios de PHP 3 (finales de los 90), esta estructura se adoptó directamente del lenguaje C,  
> lo que le dio a PHP su característica sintaxis simple y familiar para programadores con experiencia previa.


````php
<?php
$dia = "miércoles";

switch ($dia) {
    case "lunes":
        echo "Comienza la semana.";
        break;
    case "martes":
    case "miércoles":
        echo "Mitad de semana.";
        break;
    case "viernes":
        echo "¡Casi fin de semana!";
        break;
    default:
        echo "Día cualquiera.";
        break;
}
?>
````


## 11. Estructuras Repetitivas en PHP

Las **estructuras repetitivas**, también conocidas como *bucles o ciclos*, permiten ejecutar un bloque de código múltiples veces, 
ya sea mientras una condición se cumpla o hasta recorrer por completo un conjunto de elementos.

Estas estructuras son esenciales para automatizar tareas, procesar datos en masa o construir comportamientos dinámicos.

---

### 11.1. Bucle `while`

El bucle `while` ejecuta un bloque de código **mientras** la condición sea verdadera.  
Si la condición no se cumple desde el principio, el bloque **no se ejecuta ni una sola vez**.


````php

<?php
$contador = 1;

while ($contador <= 5) {
    echo "Iteración número $contador <br>";
    $contador++;
}
?>
````


---

### 11.2. Bucle `for`

El bucle `for` se utiliza cuando conocemos exactamente cuántas veces queremos que se repita un bloque de código.  
En su sintaxis, incluye tres partes: **inicialización**, **condición**, y **actualización**.


````php
<?php
for ($i = 1; $i <= 5; $i++) {
    echo "Valor actual de i: $i <br>";
}
?>
````


---

### 11.3. Bucle `foreach`

El bucle `foreach` está diseñado especialmente para recorrer **arreglos** o **colecciones de datos**.  
Es la forma más sencilla y clara de iterar sobre listas en PHP.


````php
<?php
$frutas = ["Manzana", "Banana", "Cereza"];

foreach ($frutas as $fruta) {
    echo "Fruta: $fruta <br>";
}
?>
````


## 12. Funciones en PHP

Las **funciones** son bloques de código que pueden reutilizarse en diferentes partes del programa.  
Permiten dividir el código en piezas más manejables, claras y eficientes.

PHP ofrece tanto **funciones predefinidas** como la posibilidad de crear **funciones personalizadas**.

---

### 12.1. Declaración de una función


````php

<?php
function saludar($nombre) {
    return "Hola, $nombre. Bienvenido al mundo PHP.";
}

echo saludar("Valentina");
?>
````


---

### 12.2. Funciones con valores por defecto

Puedes asignar valores por defecto a los parámetros, para que sean opcionales al momento de llamar la función.


````php

<?php
function saludar($nombre) {
    return "Hola, $nombre. Bienvenido al mundo PHP.";
}

echo saludar("Valentina");
?>
````


## 13. Manejo de Errores en PHP

El **manejo de errores** es fundamental para mantener el control del flujo del programa y evitar interrupciones inesperadas.

PHP cuenta con mecanismos nativos como `try`, `catch` y `finally` para gestionar excepciones de manera controlada.

---

### Ejemplo básico de manejo de errores


````php
<?php
try {
    $divisor = 0;
    if ($divisor == 0) {
        throw new Exception("No se puede dividir entre cero.");
    }
    echo 10 / $divisor;
} catch (Exception $e) {
    echo "Error: " . $e->getMessage();
} finally {
    echo "<br>Operación finalizada.";
}
?>
````


## 14. Arreglos en PHP

Los **arreglos** (o *arrays*) son estructuras de datos que permiten almacenar múltiples valores en una sola variable.  
PHP soporta varios tipos de arreglos: **indexados**, **asociativos** y **multidimensionales**.

---

### 14.1. Arreglos indexados


````php

<?php
$colores = ["Rojo", "Verde", "Azul"];
echo $colores[0]; // Imprime 'Rojo'
?>
````


---

### 14.2. Arreglos asociativos

Los arreglos asociativos usan **claves personalizadas** (en lugar de índices numéricos) para identificar los valores.


````php

<?php
$persona = [
    "nombre" => "Valentina",
    "edad" => 19,
    "ciudad" => "Maracay"
];

echo $persona["nombre"];
?>
````


---

### 14.3. Arreglos multidimensionales

Un arreglo puede contener otros arreglos, creando estructuras complejas útiles para representar datos jerárquicos.


````php
<?php
$estudiantes = [
    ["nombre" => "Ana", "edad" => 20],
    ["nombre" => "Luis", "edad" => 22],
    ["nombre" => "Valentina", "edad" => 19]
];

foreach ($estudiantes as $estudiante) {
    echo $estudiante["nombre"] . " tiene " . $estudiante["edad"] . " años.<br>";
}
?>
````
