## Descripción
C# (se pronuncia 'C sharp') es un lenguaje de programación moderno, orientado a objetos y desarrollado por Microsoft. Se utiliza ampliamente en desarrollo de aplicaciones de escritorio, web, videojuegos (Unity) y servicios backend con .NET.

## Instalación del Lenguaje
Para instalar C# y usarlo en Jupyter:

1. Instalar .NET SDK:
```bash
https://dotnet.microsoft.com/en-us/download
```

2. Verificar instalación:
```bash
dotnet --version
```

3. Instalar .NET Interactive (permite ejecutar C# en Jupyter):
```bash
dotnet tool install --global Microsoft.dotnet-interactive
dotnet interactive jupyter install
```

4. Luego abre Visual Studio Code y selecciona el kernel **.NET (C#)**.

## Ejemplos de sintaxis básica
Variables, tipos de datos y operaciones.

````python
// Variables en C#
string nombre = "Valentina";
int edad = 25;
double altura = 1.68;
bool esEstudiante = true;

Console.WriteLine($"Hola, mi nombre es {nombre}, tengo {edad} años y mido {altura} m.");
Console.WriteLine($"¿Soy estudiante? {esEstudiante}");
````

````python
// Tipos de datos y operaciones aritméticas
int a = 10;
int b = 3;

Console.WriteLine($"Suma: {a + b}");
Console.WriteLine($"Resta: {a - b}");
Console.WriteLine($"Multiplicación: {a * b}");
Console.WriteLine($"División: {(double)a / b}");
Console.WriteLine($"Módulo: {a % b}");
Console.WriteLine($"Potencia: {Math.Pow(a, b)}");
````

````python
// Booleanos y operadores lógicos
bool x = true;
bool y = false;

Console.WriteLine($"x AND y => {x && y}");
Console.WriteLine($"x OR y  => {x || y}");
Console.WriteLine($"NOT x   => {!x}");
````

## Estructuras de control
Condicionales y bucles.

````python
// Condicional if/else if/else
int edad = 20;

if (edad < 18)
{
    Console.WriteLine("Eres menor de edad");
}
else if (edad >= 18 && edad < 65)
{
    Console.WriteLine("Eres adulto");
}
else
{
    Console.WriteLine("Eres adulto mayor");
}
````

````python
// Bucles while y for
Console.WriteLine("\nBucle while:");
int i = 1;
while (i <= 5)
{
    Console.WriteLine(i);
    i++;
}

Console.WriteLine("\nBucle for:");
for (int j = 1; j <= 5; j++)
{
    Console.WriteLine(j);
}
````

## Funciones
Las funciones (o métodos) en C# se definen dentro de clases. 

````python
// Función simple
int Sumar(int a, int b)
{
    return a + b;
}

Console.WriteLine($"La suma es: {Sumar(5, 7)}");
````

## Programación Orientada a Objetos (POO)
C# es totalmente orientado a objetos. Se basa en clases y objetos.

````python
public class Persona
{
    public string Nombre { get; set; }
    public int Edad { get; set; }

    public Persona(string nombre, int edad)
    {
        Nombre = nombre;
        Edad = edad;
    }

    public void Saludar()
    {
        Console.WriteLine($"Hola, soy {Nombre} y tengo {Edad} años.");
    }
}

var persona1 = new Persona("Valentina", 25);
persona1.Saludar();
````

## Arreglos (vectores) y sus métodos
Los arreglos en C# son de tamaño fijo. También existen listas dinámicas (List<T>).

````python
int[] numeros = {10, 20, 30, 40, 50};

Console.WriteLine($"Primer elemento: {numeros[0]}");
Console.WriteLine($"Último elemento: {numeros[numeros.Length - 1]}");
Console.WriteLine($"Tamaño: {numeros.Length}");

Console.WriteLine("\nRecorrido del arreglo:");
foreach (int n in numeros)
{
    Console.WriteLine(n);
}

// Usando listas
var lista = new List<int> {10, 20, 30};
lista.Add(40);
lista.Remove(20);

Console.WriteLine("\nLista modificada:");
lista.ForEach(n => Console.WriteLine(n));
````
